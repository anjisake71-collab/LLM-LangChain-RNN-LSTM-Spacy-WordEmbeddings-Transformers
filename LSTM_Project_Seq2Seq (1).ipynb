```python
# -*- coding: utf-8 -*-
"""
Enhanced Character-Level Shakespeare Text Generator
Data source: "tiny_shakespeare" via TensorFlow Datasets
"""

import numpy as np
import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Embedding, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.preprocessing.text import Tokenizer
import matplotlib.pyplot as plt

# 1) Load the Tiny Shakespeare dataset from TFDS
ds = tfds.load("tiny_shakespeare", split="train", as_supervised=False)
texts = []
for example in ds:
    texts.append(example["text"].numpy().decode("utf-8"))
text = "\n".join(texts)  # Keep original case for more Shakespearean feel
print(f"Total characters: {len(text)}")

# 2) Character-level tokenization
tokenizer = Tokenizer(char_level=True, oov_token=None)
tokenizer.fit_on_texts(text)
vocab_size = len(tokenizer.word_index) + 1
print(f"Vocabulary size: {vocab_size}")

# 3) Create input sequences of length 100 + 1 target (longer for better context)
sequences = []
seq_length = 100  # Increased from 40 to 100 for better context
for i in range(seq_length, len(text)):
    seq = text[i - seq_length : i + 1]
    encoded = [tokenizer.texts_to_sequences([c])[0][0] for c in seq]
    sequences.append(encoded)
sequences = np.array(sequences)
print(f"Number of sequences: {len(sequences)}")

# 4) Split into X (inputs) and y (targets)
X = sequences[:, :-1]
y = sequences[:, -1]

# 5) One-hot encode the y (targets)
y = tf.keras.utils.to_categorical(y, num_classes=vocab_size)

# 6) Build a more powerful LSTM model with multiple layers
model = Sequential([
    Embedding(vocab_size, 128, input_length=seq_length),  # Embedding instead of one-hot
    LSTM(256, return_sequences=True),  # First LSTM layer with more units
    Dropout(0.2),  # Prevent overfitting
    LSTM(256),  # Second LSTM layer
    BatchNormalization(),  # Stabilize training
    Dense(256, activation='relu'),  # Additional dense layer
    Dropout(0.2),  # More dropout
    Dense(vocab_size, activation='softmax')  # Output layer
])

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
model.summary()

# 7) Add callbacks for better training
callbacks = [
    EarlyStopping(patience=5, restore_best_weights=True),
    ReduceLROnPlateau(factor=0.5, patience=2)
]

# 8) Train the model with validation split
history = model.fit(
    X, y,
    epochs=50,
    batch_size=128,  # Larger batch size
    validation_split=0.1,  # Use 10% for validation
    callbacks=callbacks
)

# 9) Plot training loss and accuracy
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='train_loss')
plt.plot(history.history['val_loss'], label='val_loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='train_acc')
plt.plot(history.history['val_accuracy'], label='val_acc')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# 10) Generate text with temperature sampling
def generate_text(seed_text, num_chars=400, temperature=0.7):
    """Generate text with temperature sampling for more creative output"""
    generated_text = seed_text

    # Make sure seed text is exactly seq_length characters
    if len(seed_text) < seq_length:
        seed_text = seed_text.rjust(seq_length)  # Pad with spaces if needed
    seed_text = seed_text[-seq_length:]  # Take last seq_length characters

    for _ in range(num_chars):
        # Convert characters to sequences
        encoded = [tokenizer.texts_to_sequences([c])[0][0] for c in seed_text]

        # Predict next character with temperature
        preds = model.predict(np.array([encoded]), verbose=0)[0]

        # Apply temperature sampling
        preds = np.log(preds) / temperature
        exp_preds = np.exp(preds)
        preds = exp_preds / np.sum(exp_preds)

        # Sample from the distribution
        next_index = np.random.choice(vocab_size, p=preds)
        next_char = tokenizer.index_word.get(next_index, '')

        # Update seed text and generated text
        generated_text += next_char
        seed_text = seed_text[1:] + next_char  # Remove first char and add new char at end

    return generated_text

# 11) Generate text with different temperatures
print("\nGenerated Text (temperature=0.5, more focused):\n")
print(generate_text("Shall I compare thee to a summer's day?", temperature=0.5))

print("\nGenerated Text (temperature=0.7, balanced):\n")
print(generate_text("Shall I compare thee to a summer's day?", temperature=0.7))

print("\nGenerated Text (temperature=1.0, more creative):\n")
print(generate_text("Shall I compare thee to a summer's day?", temperature=1.0))
```
