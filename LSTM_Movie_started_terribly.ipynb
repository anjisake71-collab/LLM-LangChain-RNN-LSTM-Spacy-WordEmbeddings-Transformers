```python
# Import necessary libraries
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

# 1. Extended Dataset
sentences = [
    "The movie started terribly, but the ending was absolutely amazing!",
    "The acting was horrible and the storyline was dull.",
    "I loved the cinematography and the direction was brilliant!",
    "The film started slow and ended on a disappointing note.",
    "What an amazing performance! The entire movie was a delight.",
    "It was a waste of time, completely boring and poorly acted.",
    "The movie was bad at first, but ended beautifully.",
    "The visuals were stunning, and the plot was engaging.",
    "Terrible movie. I regret watching it.",
    "Fantastic storyline with brilliant execution!"
]

# Labels: 1 - Positive, 0 - Negative
labels = [1, 0, 1, 0, 1, 0, 1, 1, 0, 1]

# 2. Tokenization and Padding
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts(sentences)
sequences = tokenizer.texts_to_sequences(sentences)
padded_sequences = pad_sequences(sequences, padding='post')

# Print tokenized sentences
print("Vocabulary: ", tokenizer.word_index)

# 3. Updated LSTM Model
vocab_size = len(tokenizer.word_index) + 1
embedding_dim = 32  # Increased embedding dimension
max_length = len(padded_sequences[0])

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    LSTM(64),  # Increased LSTM units
    Dense(1, activation='sigmoid')  # Binary classification
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# 4. Train the Model
labels = np.array(labels)
model.fit(padded_sequences, labels, epochs=20, verbose=2)  # Increased epochs

# 5. Test the Model on a New Sentence
new_sentence = ["The movie started terribly, but the ending was absolutely amazing!"]
new_sequence = tokenizer.texts_to_sequences(new_sentence)
new_padded_sequence = pad_sequences(new_sequence, maxlen=max_length, padding='post')

prediction = model.predict(new_padded_sequence)
sentiment = "Positive" if prediction[0][0] > 0.5 else "Negative"

print(f"\nInput Sentence: {new_sentence[0]}")
print(f"Predicted Sentiment: {sentiment}")
```
