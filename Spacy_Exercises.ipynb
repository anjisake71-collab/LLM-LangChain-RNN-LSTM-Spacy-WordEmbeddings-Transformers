```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
print("spaCy model loaded successfully!")
```

```python
sample_text = """
Hello! I'm learning about spaCy in 2024. It's a POWERFUL library for Natural Language Processing, isn't it?
Dr. Smith works at Google LLC in New York. She earned $100,000 last year.
The company's website is https://www.example.com. Contact: info@example.com
Some numbers: 42, 3.14, 1st, 2nd, 100%.
"""
```

```python
doc = nlp(sample_text)
```

```python
for token in doc:
    print(token.text)
```

**Basic Tokenization**

```python
tokens = [token.text for token in doc]
print(f"Total tokens: {len(tokens)}")
print("First 20 tokens:", tokens[:20])
```

**Lowercasing**

```python
lowercase_tokens = [token.text.lower() for token in doc]
print("Original:", [token.text for token in doc][:10])
print("Lowercase:", lowercase_tokens[:10])
```

**Remove Punctuation**

```python
no_punct = [token.text for token in doc if not token.is_punct]
print(f"With punctuation: {len(tokens)} tokens")
print(f"Without punctuation: {len(no_punct)} tokens")
```

**Remove Stop words**

```python
no_stop = [token.text for token in doc if not token.is_stop]
print(f"With stop words: {len(tokens)} tokens")
print(f"Without stop words: {len(no_stop)} tokens")
```

**Remove Punctuation AND Stop Words**

```python
filtered = [token.text for token in doc if not token.is_stop and not token.is_punct]
print(f"Original: {len(tokens)} tokens")
print(f"Filtered: {len(filtered)} tokens")
print("Result:", filtered)
```

**Lemmatization**

```python
lemmas = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
print("Original word -> Base form (lemma):")
for token in doc[:15]:
    if not token.is_stop and not token.is_punct:
        print(f"{token.text:15} -> {token.lemma_}")
```

**Remove Numbers/Digits**

```python
no_numbers = [token.text for token in doc if not token.is_digit and not token.like_num]
print("Numbers found:", [token.text for token in doc if token.is_digit or token.like_num])
print(f"After removing numbers: {len(no_numbers)} tokens")
```

**Part-of-Speech (POS) Tagging**

```python
for token in doc[:20]:
    if not token.is_space:
        print(f"{token.text:15} | {token.pos_:8} | {token.tag_}")
```

**Named Entity Recognition (NER)**

```python
for ent in doc.ents:
    print(f"{ent.text:25} | {ent.label_:13} | {spacy.explain(ent.label_)}")
```

**Sentence Segmentation**

```python
sentences = [sent.text.strip() for sent in doc.sents]
print(f"Total sentences: {len(sentences)}\n")
for i, sent in enumerate(sentences, 1):
    print(f"Sentence {i}: {sent}")
```

**Dependency Parsing**

```python
for token in doc[:25]:
    if not token.is_space:
        print(f"{token.text:15} | {token.dep_:12} | {token.head.text}")
```

**Complete Preprocessing Pipeline**

```python
def preprocess_text(text):
    """
    Complete preprocessing pipeline:
    1. Tokenization
    2. Lowercase
    3. Remove punctuation
    4. Remove stop words
    5. Remove numbers
    6. Remove URLs/emails
    7. Lemmatization
    """
    doc = nlp(text)

    processed = [
        token.lemma_.lower()
        for token in doc
        if not token.is_stop
        and not token.is_punct
        and not token.is_space
        and not token.like_num
        and not token.like_url
        and not token.like_email
    ]

    return processed

preprocessed = preprocess_text(sample_text)
print(f"Preprocessed result ({len(preprocessed)} tokens):")
print(preprocessed)

# ========================================
# CELL 20: Extract Only Alphabetic Tokens
# Task: Keep only pure alphabetic tokens
# ========================================
alpha_only = [token.text for token in doc if token.is_alpha]
print("Alphabetic tokens only:", alpha_only[:20])

# ========================================
# CELL 21: Custom Stop Words
# Task: Add or remove custom stop words
# ========================================

# View some default stop words
print("Sample stop words:", list(nlp.Defaults.stop_words)[:10])

# Add custom stop word
nlp.Defaults.stop_words.add("spacy")
nlp.vocab["spacy"].is_stop = True

# Remove a stop word
nlp.Defaults.stop_words.remove("not")
nlp.vocab["not"].is_stop = False

print("\nAfter customization:")
print("'spacy' is stop word:", nlp.vocab["spacy"].is_stop)
print("'not' is stop word:", nlp.vocab["not"].is_stop)

# ========================================
# CELL 22: N-grams (Bigrams and Trigrams)
# Task: Extract word sequences
# ========================================

def get_ngrams(doc, n=2):
    """Extract n-grams from doc"""
    tokens = [token.text.lower() for token in doc if not token.is_stop and not token.is_punct]
    ngrams = [' '.join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]
    return ngrams

bigrams = get_ngrams(doc, n=2)
trigrams = get_ngrams(doc, n=3)

print("Bigrams (first 10):", bigrams[:10])
print("Trigrams (first 10):", trigrams[:10])

# ========================================
# CELL 23: Frequency Analysis
# Task: Count token frequencies
# ========================================
from collections import Counter

clean_tokens = [
    token.lemma_.lower()
    for token in doc
    if not token.is_stop and not token.is_punct and token.is_alpha
]

freq = Counter(clean_tokens)
print("Most common tokens:")
for token, count in freq.most_common(10):
    print(f"{token:15} : {count}")

# ========================================
# CELL 24: Token Length Filter
# Task: Filter tokens by length
# ========================================

long_tokens = [token.text for token in doc if len(token.text) >= 3 and token.is_alpha]
print(f"Tokens with 3+ characters: {len(long_tokens)}")
print("Sample:", long_tokens[:15])

# ========================================
# CELL 25: Save Preprocessed Text
# Task: Join processed tokens back into text
# ========================================

final_processed = preprocess_text(sample_text)
final_text = ' '.join(final_processed)

print("Original text length:", len(sample_text), "characters")
print("Processed text length:", len(final_text), "characters")
print("\nFinal preprocessed text:")
print(final_text)
```
