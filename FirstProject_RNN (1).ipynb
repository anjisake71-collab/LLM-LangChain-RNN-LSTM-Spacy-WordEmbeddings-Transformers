```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# 1️⃣ Load and preprocess IMDB dataset
max_features = 10000  # only consider the top 10K words in the dataset
maxlen = 500          # cut reviews after 500 words
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

x_train = sequence.pad_sequences(x_train, maxlen=maxlen)
x_test  = sequence.pad_sequences(x_test, maxlen=maxlen)

# 2️⃣ Build the Simple RNN model
model = Sequential([
    Embedding(input_dim=max_features, output_dim=32, input_length=maxlen),
    SimpleRNN(32),  # classic RNN layer for sequence processing
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
model.summary()

# 3️⃣ Train the model
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=64,
    validation_data=(x_test, y_test),
    verbose=1
)

# 4️⃣ Evaluate model performance
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f"\n✅ Test Accuracy: {test_acc:.2f}")

# 5️⃣ Optional: Predict sentiment on new raw text
from tensorflow.keras.preprocessing.text import Tokenizer

# Build tokenizer with same vocabulary size
tokenizer = Tokenizer(num_words=max_features)
word_index = imdb.get_word_index()
tokenizer.word_index = {k:(v+3) for k,v in word_index.items()}
tokenizer.word_index["<PAD>"] = 0
tokenizer.word_index["<START>"] = 1
tokenizer.word_index["<UNK>"] = 2
tokenizer.word_index["<UNUSED>"] = 3

def preprocess_text(texts):
    sequences = tokenizer.texts_to_sequences(texts)
    return sequence.pad_sequences(sequences, maxlen=maxlen)

def predict_sentiment(texts):
    seq = preprocess_text(texts)
    preds = model.predict(seq)
    return ["Positive" if p > 0.8 else "Negative" for p in preds]

# Example predictions
samples = [
    "I absolutely loved this movie! It was thrilling.",
    "It was so boring and too long. I hated it."
]
print("\nSample Predictions:")
for text, sentiment in zip(samples, predict_sentiment(samples)):
    print(f"» “{text}” → {sentiment}")
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.preprocessing.text import Tokenizer

max_features = 10000
maxlen = 500

# Load and preprocess IMDB
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)
x_train = sequence.pad_sequences(x_train, maxlen=maxlen)
x_test  = sequence.pad_sequences(x_test, maxlen=maxlen)

# Build model
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense

model = Sequential([
    Input(shape=(maxlen,)),
    Embedding(input_dim=max_features, output_dim=32),
    SimpleRNN(32),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))
print(f"Test Accuracy: {model.evaluate(x_test, y_test, verbose=2)[1]:.2f}")

# Tokenizer consistent with IMDB indices
word_index = imdb.get_word_index()
tokenizer = Tokenizer(num_words=max_features)
tokenizer.word_index = {k:(v+3) for k,v in word_index.items()}
tokenizer.word_index["<PAD>"] = 0
tokenizer.word_index["<START>"] = 1
tokenizer.word_index["<UNK>"] = 2
tokenizer.word_index["<UNUSED>"] = 3

def encode_review(text):
    tokens = [1]
    for w in text.lower().split():
        idx = tokenizer.word_index.get(w, 2)
        if idx >= max_features:
            idx = 2
        tokens.append(idx)
    return sequence.pad_sequences([tokens], maxlen=maxlen)

def predict_sentiment(texts):
    seqs = [encode_review(t)[0] for t in texts]
    preds = model.predict(tf.constant(seqs))
    return ["Positive" if p > 0.5 else "Negative" for p in preds]

custom_reviews = [
    "The movie was inspiring and had a strong ending",
    "It dragged and I regretted watching it till the end",
    "Not fantastic, though some scenes were enjoyable",
    "Impressive visuals but the dialogue was flat",
    "It had high expectations but delivered too little"
]

print("\nCustom Predictions:")
for txt, out in zip(custom_reviews, predict_sentiment(custom_reviews)):
    print(f"“{txt}” → {out}")
```

**BBC News**

```python
import pandas as pd
import numpy as np
import re
import nltk
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelBinarizer

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping

# --- 1. Load and Prepare Data ---

# Download stopwords if you haven't already
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Load the dataset
# Make sure 'bbc-text.csv' is in the same directory or provide the correct path
try:
    df = pd.read_csv('bbc-text.csv')
except FileNotFoundError:
    print("Error: 'bbc-text.csv' not found. Please download it from Kaggle and place it in the same directory.")
    exit()

# --- 2. Text Preprocessing ---

# Define english stopwords
stop_words = set(stopwords.words('english'))

def clean_text(text):
    """
    Cleans the input text by lowercasing, removing punctuation,
    numbers, and stopwords.
    """
    text = text.lower()  # Lowercase text
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    tokens = text.split()
    # Remove stopwords
    clean_tokens = [word for word in tokens if word not in stop_words]
    return " ".join(clean_tokens)

# Apply the cleaning function
print("Cleaning text data...")
df['clean_text'] = df['text'].apply(clean_text)

# --- 3. Prepare Data for the Model ---

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df['clean_text'],
    df['category'],
    test_size=0.2,
    random_state=42,
    stratify=df['category'] # Ensures balanced split
)

# Vectorize text using Keras Tokenizer
vocab_size = 10000  # Max number of words to keep
embedding_dim = 128
max_length = 200    # Max length of sequences

tokenizer = Tokenizer(num_words=vocab_size, oov_token='<OOV>')
tokenizer.fit_on_texts(X_train)

# Convert text to sequences of integers
X_train_sequences = tokenizer.texts_to_sequences(X_train)
X_test_sequences = tokenizer.texts_to_sequences(X_test)

# Pad sequences to ensure uniform length
X_train_padded = pad_sequences(X_train_sequences, maxlen=max_length, padding='post', truncating='post')
X_test_padded = pad_sequences(X_test_sequences, maxlen=max_length, padding='post', truncating='post')

# Binarize the labels (e.g., 'sport' -> [0,0,0,0,1])
label_binarizer = LabelBinarizer()
y_train_final = label_binarizer.fit_transform(y_train)
y_test_final = label_binarizer.transform(y_test)
num_classes = len(label_binarizer.classes_)

# --- 4. Build and Train the SimpleRNN Model ---

print("\nBuilding the model...")
model = Sequential([
    # Input Layer: Learns word embeddings
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_length),
    # Hidden Layer: Bidirectional SimpleRNN processes sequences
    Bidirectional(SimpleRNN(64)),
    # Output Layer: Classifies into one of the categories
    Dense(num_classes, activation='softmax')
])

# Compile the model
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# Set up early stopping to prevent overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# Train the model
print("\nTraining the model...")
history = model.fit(
    X_train_padded,
    y_train_final,
    epochs=4,
    batch_size=32,
    validation_split=0.1,
    callbacks=[early_stopping]
)

# --- 5. Evaluate the Model ---

print("\nEvaluating the model on the test set...")
loss, accuracy = model.evaluate(X_test_padded, y_test_final)
print(f"Test Accuracy: {accuracy * 100:.2f}%")

# --- 6. Prediction on New Data ---

def predict_category(text, model, tokenizer, label_binarizer):
    """Predicts the category of a new piece of text."""
    # Clean and preprocess the new text
    cleaned_text = clean_text(text)
    sequence = tokenizer.texts_to_sequences([cleaned_text])
    padded_sequence = pad_sequences(sequence, maxlen=max_length, padding='post', truncating='post')

    # Make a prediction
    prediction = model.predict(padded_sequence)
    predicted_class_index = np.argmax(prediction, axis=1)[0]

    # Return the category name
    return label_binarizer.classes_[predicted_class_index]


# Example usage
print("\n--- Making a new prediction ---")
new_article = """
The government announced new fiscal policies today aimed at curbing inflation
and promoting investment in the technology sector. The stock market responded
positively to the news, with major indices showing significant gains.
"""

predicted_category = predict_category(new_article, model, tokenizer, label_binarizer)
print(f"The predicted category for the new article is: '{predicted_category}'")
```
